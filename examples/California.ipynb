# Building the California example graph

Load the WECC graph, keep the six California balancing regions, and aggregate similar assets.
Aggregation only merges plants of the same class, type, fuel, profile and jurisdiction, and it
preserves total capacity.

```python
import good
```

```python
graph = good.graph.graph_from_json('data/WEC.json.gz')

nodes = ['WEC_BANC', 'WEC_CALN', 'WEC_LADW', 'WEC_SDGE', 'WECC_IID', 'WECC_SCE']
graph = good.graph.subgraph(graph, nodes)

print(f"{sum(len(d['assets']) for _, d in graph.nodes(data=True))} assets before aggregation")
```

`ratio` sets the number of clusters per group of similar assets as a fraction of the group's size;
`max_clusters` caps it. Clustering uses k-means on heat rate, operating cost and CO2 rate.

```python
graph = good.aggregate.aggregate(graph, ratio=0.1)

print(f"{sum(len(d['assets']) for _, d in graph.nodes(data=True))} assets after aggregation")
```

```python
import os

os.makedirs('outputs', exist_ok=True)
good.graph.graph_to_json(graph, 'outputs/California_aggregated.json.gz')
```
